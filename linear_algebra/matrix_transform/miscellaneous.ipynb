```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 120)
y = np.linspace(-4, 4, 120)
z = np.linspace(-4, 4, 120)

X, Y, Z = np.meshgrid(x,y,z)

F = (
    np.sin(X**2 + Y**2)
    + np.cos(Y**2 + Z**2)
    + np.sin(Z**2 + X**2)
    + 0.3*np.sin(X*Y*Z)
)
```

```python
print(X.shape)
print(X.ndim)
print()
print(Y.shape)
print()

a = np.sin(X**2 + Y**2)
print(a.shape)
```

```python
import pandas as pd

df = pd.DataFrame({

    "x": X.ravel(),
    "y": Y.ravel(),
    "z": Z.ravel(),
    "value": F.ravel()

})

print(df.head())
```
