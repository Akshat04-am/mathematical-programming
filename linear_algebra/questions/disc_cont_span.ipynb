###  For v=(1,0) and w=(0,1), describe all points cv with.

1. whole numbers c

2.  nonnegative numbers c≥0. Then add all vectors dw and describe all cv+dw.

```python
import matplotlib.pyplot as plt
import numpy as np

# whole-number values of c
c_discrete = np.arange(0, 16)

# continuous values of c
c_cont = np.linspace(0, 15, 200)

# Since v = (1,0)
# cv = (c,0)

x_discrete = c_discrete
y_discrete = np.zeros_like(c_discrete)

x_cont = c_cont
y_cont = np.zeros_like(c_cont)

# Plot
plt.scatter(x_discrete, y_discrete, label="whole numbers")
# plt.plot(x_discrete, y_discrete, label="whole numbers")
plt.plot(x_cont, y_cont, label="continuous")

# plt.axhline(0)
# plt.axvline(0)

plt.grid(True)
plt.legend()
plt.axis("equal")

plt.show()
```

### Combined but gets advanced.

```python
import numpy as np
import matplotlib.pyplot as plt

# v = (1,0), w = (0,1)
v = np.array([1, 0])
w = np.array([0, 1])

# -----------------------------
# 1) cv with whole numbers c
# -----------------------------
c_whole = np.arange(0, 11)              # 0,1,2,...,10
points_whole = np.column_stack((c_whole, np.zeros_like(c_whole)))  # (c,0)

# -----------------------------
# 2) cv with nonnegative c >= 0
# -----------------------------
c_nonneg = np.linspace(0, 10, 200)      # many nonnegative values
points_nonneg = np.column_stack((c_nonneg, np.zeros_like(c_nonneg)))  # (c,0)

# -----------------------------
# 3) cv + dw
#    Since v=(1,0), w=(0,1):
#    cv + dw = (c,0) + (0,d) = (c,d)
# -----------------------------
c = np.arange(0, 11)                    # whole-number c
d = np.arange(0, 11)                    # whole-number d
C, D = np.meshgrid(c, d)

points_combined = np.column_stack((C.ravel(), D.ravel()))  # all (c,d) pairs

# -----------------------------
# Plot
# -----------------------------
fig, axs = plt.subplots(1, 3, figsize=(15, 4))

# Part 1
axs[0].scatter(points_whole[:, 0], points_whole[:, 1])
axs[0].set_title("cv, c = whole numbers")
# axs[0].axhline(0)
# axs[0].axvline(0)
axs[0].grid(True)
axs[0].axis("equal")

# Part 2
axs[1].plot(points_nonneg[:, 0], points_nonneg[:, 1])
axs[1].set_title("cv, c >= 0")
# axs[1].axhline(0)
# axs[1].axvline(0)
axs[1].grid(True)
axs[1].axis("equal")

# Part 3
axs[2].scatter(points_combined[:, 0], points_combined[:, 1])
axs[2].set_title("cv + dw, c and d whole numbers")
# axs[2].axhline(0)
# axs[2].axvline(0)
axs[2].grid(True)
axs[2].axis("equal")

plt.tight_layout()
plt.show()
```
