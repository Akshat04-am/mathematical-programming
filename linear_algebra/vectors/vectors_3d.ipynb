```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

#  Create figure
fig = plt.figure(figsize=(6,6))

#  Create 3D axis
ax = fig.add_subplot(111, projection='3d')

# ==================================
# VECTOR
# ==================================

v = (2, 1, 3)

# ==================================
# DRAW VECTOR
# ==================================

#  Start point = (0,0,0)
#  Direction = (2,1,3)

ax.quiver(
    0, 0, 0,     # start point
    v[0], v[1], v[2]   # vector direction
)

# ==================================
# AXIS LIMITS
# ==================================

for setter in [ax.set_xlim, 
               ax.set_ylim, 
               ax.set_zlim]:
    setter([0, 5])

# ==================================
# LABELS
# ==================================

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

#  Create figure
fig = plt.figure(figsize=(7,7))

#  Create 3D axes
ax = fig.add_subplot(111, projection='3d')

# ==================================
# VECTORS
# ==================================

v = (2, 1, 3)
w = (1, 3, 1)

# ==================================
# DRAW ORIGINAL VECTORS
# ==================================

# v from origin
ax.quiver(0, 0, 0,
          v[0], v[1], v[2])

# w from origin
ax.quiver(0, 0, 0,
          w[0], w[1], w[2])

# ==================================
# PARALLELOGRAM COPIES
# ==================================

#  Copy w to tip of v
ax.quiver(v[0], v[1], v[2],
          w[0], w[1], w[2])

#  Copy v to tip of w
ax.quiver(w[0], w[1], w[2],
          v[0], v[1], v[2])

# ==================================
# RESULT VECTOR
# ==================================

result = (
    v[0] + w[0],
    v[1] + w[1],
    v[2] + w[2]
)

#  Diagonal vector
ax.quiver(0, 0, 0,
          result[0], result[1], result[2])

# ==================================
# AXIS LIMITS
# ==================================

ax.set_xlim([0, 5])
ax.set_ylim([0, 5])
ax.set_zlim([0, 5])

# Labels
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show()
```
