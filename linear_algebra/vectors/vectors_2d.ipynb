1. **For 2D vectors**

```python
import matplotlib.pyplot as plt

# vector = (3, 2)
x,y = 3,2

#  Draw vector from origin (0,0)
plt.quiver(0, 0, x, y, 
           angles='xy', scale_units='xy', scale=1)

plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.grid()
plt.show()
```

2. **Multiple vectors**

```python
import matplotlib.pyplot as plt

vectors = [
    (3, 2),
    (-2, 1),
    (1, -3)
]

for x, y in vectors:
    
    #  Draw each vector from origin
    plt.quiver(0, 0, x, y,
               angles='xy', scale_units='xy', scale=1)

plt.xlim(-5, 5)
plt.ylim(-5, 5)
plt.grid()
plt.show()
```

3. **Vector addition**

```python
import matplotlib.pyplot as plt

v = (2, 1)
w = (1, 3)

#  Add vectors component-wise
result = (v[0] + w[0], v[1] + w[1])

# v
plt.quiver(0, 0, v[0], v[1],
           angles='xy', scale_units='xy', scale=1)

# w
plt.quiver(0, 0, w[0], w[1],
           angles='xy', scale_units='xy', scale=1)

# v + w
plt.quiver(0, 0, result[0], result[1],
           angles='xy', scale_units='xy', scale=1)

plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.grid()
plt.show()
```

4. **Parallelogram rule for vector addition.**

```python
import matplotlib.pyplot as plt

#  Original vectors
v = (2, 1)
w = (1, 3)

#  Vector addition
result = (v[0] + w[0], v[1] + w[1])

# ORIGINAL VECTORS

# v from origin
plt.quiver(0, 0, v[0], v[1],
           angles='xy', scale_units='xy', scale=1)

# w from origin
plt.quiver(0, 0, w[0], w[1],
           angles='xy', scale_units='xy', scale=1)


# DUPLICATED VECTORS

#  Copy w to tip of v
plt.quiver(v[0], v[1], w[0], w[1],
           angles='xy', scale_units='xy', scale=1)

#  Copy v to tip of w
plt.quiver(w[0], w[1], v[0], v[1],
           angles='xy', scale_units='xy', scale=1)


# RESULT VECTOR
# Diagonal = v + w
plt.quiver(0, 0, result[0], result[1],
           angles='xy', scale_units='xy', scale=1)

plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.grid()
plt.show()
```
