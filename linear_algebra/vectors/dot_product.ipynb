1. **Calculating dot product or  numerator**

```python
import numpy as np

a = np.array([1, 2])
b = np.array([4, 5])

dot_product = np.dot(a, b)
print(dot_product)
```

```python
vec_product1 = np.hypot(a[0], a[1])
vec_product2 = np.hypot(b[0], b[1])
print(f"Vector Product 1: {vec_product1.round(2)}")
print(f"Vector Product 2: {vec_product2.round(2)}")

both_product = vec_product1 * vec_product2
print(f"Combined Product: {both_product.round(2)}")
```

```python
cos_result = np.cos(dot_product/both_product).round(2)

print(cos_result)
```

```python
import numpy as np

# vectors
a = np.array([1, 2])
b = np.array([4, 5])

#dot product.
dot_product = np.dot(a, b)
print(f"Dot Product: {dot_product}")


# combined vector-length product.
vec_product1 = np.hypot(a[0], a[1])
vec_product2 = np.hypot(b[0], b[1])
print(f"Vector Product 1: {vec_product1.round(2)}")
print(f"Vector Product 2: {vec_product2.round(2)}")

both_product = vec_product1 * vec_product2
print(f"Combined Product: {both_product.round(2)}")


# cos result.
cos_theta = (dot_product / both_product).round(2)

print(f"Cos Result: {cos_theta}")
```
