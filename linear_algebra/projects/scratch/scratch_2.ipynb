```python
"""Visualizing how each matrix column contributes
to the final transformed vector."""

import numpy as np
import matplotlib.pyplot as plt

# basis/transformation columns
x = np.array([[1],
              [2]])

y = np.array([[3],
              [2]])

# transformation matrix
matrix = np.column_stack((x, y))

det = np.linalg.det(matrix)

print("Matrix:\n", matrix)
print("Determinant:", det)

# input vector
vector = np.array([[1],
                   [5]])

# individual column contributions
part_1 = vector[0,0] * matrix[:,0]
part_2 = vector[1,0] * matrix[:,1]

# final transformed vector
result = part_1 + part_2

# original vector
plt.quiver(
    0, 0,
    vector[0,0], vector[1,0],
    angles='xy',
    scale_units='xy',
    scale=1,
    label="original vector"
)

# first column contribution
plt.quiver(
    0, 0,
    part_1[0], part_1[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color="red",
    label="column 1 contribution"
)

# second column contribution
plt.quiver(
    0, 0,
    part_2[0], part_2[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color="yellow",
    label="column 2 contribution"
)

# final transformed vector
plt.quiver(
    0, 0,
    result[0], result[1],
    angles='xy',
    scale_units='xy',
    scale=1,
    color="green",
    label="final transformed vector"
)

plt.xlim(0,20)
plt.ylim(0, 20)

plt.axhline(0, color='red')
plt.axvline(0, color='red')

plt.grid()
plt.legend()
plt.show()
```

```python
import numpy as np

# Your current setup: A basic 2D shear matrix
A = np.array([[1, 2],
              [0, 1]])

# A simple grid of vectors
x = np.linspace(-2, 2, 5)
y = np.linspace(-2, 2, 5)
X, Y = np.meshgrid(x, y)
vectors = np.vstack([X.ravel(), Y.ravel()])

# Apply transformation
transformed = A @ vectors

# --- THE FLUENCY CHECK (No advanced concepts needed) ---
# Pick one explicit basis vector from your grid to test, like [0, 1]^T
# According to linear algebra, tracking where the basis vectors land tells you everything.
# For our matrix A, the vector [0, 1]^T must land exactly at [2, 1]^T.

# Find a vector in your grid where x=0 and y=1
idx = np.where((vectors[0] == 0) & (vectors[1] == 1))[0][0]

original_v = vectors[:, idx]
transformed_v = transformed[:, idx]

print(f"Original basis vector: {original_v}")
print(f"Transformed basis vector lands at: {transformed_v}")

# Raw, simple validation: Force the execution layer to verify the math
if np.allclose(transformed_v, [2, 1]):
    print("Intuition confirmed: Basis vector landed exactly where expected. The plot is mathematically true.")
else:
    print("Error: The transformation matrix or multiplication order is altering space incorrectly.")
```
