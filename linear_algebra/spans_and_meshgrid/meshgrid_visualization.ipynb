```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 120)
y = np.linspace(-4, 4, 120)
z = np.linspace(-4, 4, 120)

X, Y, Z = np.meshgrid(x, y, z)

F = (
    np.sin(X**2 + Y**2)
    + np.cos(Y**2 + Z**2)
    + np.sin(Z**2 + X**2)
    + 0.3*np.sin(X*Y*Z)
)

mid = len(z) // 2

slice_2d = F[:, :, mid]

plt.figure(figsize=(10,8))
plt.imshow(
    slice_2d,
    extent=[x.min(), x.max(), y.min(), y.max()],
    origin='lower'
)
plt.colorbar(label='Field Intensity')
plt.title("Slice Through A 3D Scalar Field")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Build a 3D grid
# -----------------------------
x = np.linspace(-4, 4, 120)
y = np.linspace(-4, 4, 120)
z = np.linspace(-4, 4, 120)

X, Y, Z = np.meshgrid(x, y, z, indexing="ij")

# -----------------------------
# Define one 3D scalar field
# -----------------------------
F = (
    np.sin(X**2 + Y**2)
    + np.cos(Y**2 + Z**2)
    + np.sin(Z**2 + X**2)
    + 0.3 * np.sin(X * Y * Z)
)

# -----------------------------
# Choose different Z-slices
# -----------------------------
slice_indices = [10, 35, 60, 85]

fig, axes = plt.subplots(1, 4, figsize=(18, 4), constrained_layout=True)

for ax, k in zip(axes, slice_indices):
    slice_2d = F[:, :, k]   # fixed z = z[k], vary x and y

    im = ax.imshow(
        slice_2d,
        extent=[x.min(), x.max(), y.min(), y.max()],
        origin="lower",
        aspect="auto"
    )

    ax.set_title(f"z = {z[k]:.2f}")
    ax.set_xlabel("x")
    ax.set_ylabel("y")

fig.colorbar(im, ax=axes, shrink=0.9, label="Field intensity")
plt.suptitle("Different Z-slices of the same 3D field", y=1.05)
plt.show()
```
