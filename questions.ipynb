Use these. They are narrow, messy, and actually worth the effort.

1. **What happens to (\log(1+x)) near (x=0)?**
   Compare NumPy evaluation to the first few terms of the Taylor expansion, and look for cancellation.

2. **How does (\frac{(x+1)-1}{x}) behave near (x=0)?**
   This is the classic precision trap: algebra says 1, floats may disagree.

3. **What does (\frac{x^2-1}{x-1}) look like near (x=1)?**
   Good removable discontinuity test. Exact simplification versus raw numerical evaluation.

4. **How stable is (e^x-1) for very small (x)?**
   Compare against SymPy and see where direct evaluation loses precision.

5. **How does (\sin(x)/x) behave near (x=0)?**
   Great for domain clustering and limit behavior.

6. **What happens to (\frac{1-\cos(x)}{x^2}) near (x=0)?**
   Another cancellation-heavy expression that reveals floating-point weakness fast.

7. **How does vertical shifting affect (f(x)=x^2), (x^3), or (\sqrt{x})?**
   Simple structurally, but good for building clean vectorization habits.

8. **What changes when you apply a horizontal shift to (\log(x)) or (\frac{1}{x})?**
   Forces you to track domain restrictions properly.

9. **What does matrix multiplication do to shape compatibility for (2\times 3) and (3\times 2) arrays?**
   Good for basic array discipline before you drift into theory.

10. **How does repeated composition behave for (f(x)=\sqrt{x}) or (f(x)=\log(x))?**
    Domain shrinkage becomes obvious fast.

Best first pick: **(\frac{x^2-1}{x-1}) near (x=1)** or **(\log(1+x)) near (x=0)**.
Those give you the cleanest Sandpit → Cleanroom → Diff cycle.
