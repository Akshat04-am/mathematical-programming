### Vision
Programming is the **execution layer** of your mathematics. It is a tool for building a "Numerical Laboratory" where static formulas are transformed into dynamic, vectorized systems. The goal is to move beyond rote calculation and develop a deep, first-principles intuition for how variables interact at scale.

---

### Modus Operandi (MO)

*   **No Theory Debt:** Programming sessions are for implementation and verification, not for learning new mathematical theory. You only code concepts you have already mastered in your textbook.
*   **Vector-First Implementation:** Avoid scalar thinking and Python loops. Every problem must be modeled as a set or a distribution using **NumPy** arrays to mirror the behavior of mathematical functions across a domain.
*   **Logic over Structure:** Prioritize the mathematical transformation ($f(x)$) over data organizational "bloat." Use raw NumPy and standard Matplotlib to visualize behavior. Complex data handling (like Tidy Data/Melt) is secondary to the correctness of the logic.
*   **The Truth Engine (Dual-Track Verification):** Use **SymPy** for symbolic/algebraic exactness and **NumPy** for numerical simulation. Verify that your numerical results converge toward the symbolic truth.
*   **Stress-Testing:** Actively look for the "breaking points." Use computation to observe how functions behave at limits ($x \to 0$ or $x \to \infty$) and identify where the computer's floating-point precision fails.

---

### The Logical Progression

1.  **Formula Recall:** Establish the formula and its properties based on master notes.
2.  **Vectorized Build:** Construct the domain using `np.linspace` or `np.logspace` and define the transformation logic.
3.  **Verification:** Compare numerical output against symbolic solutions or manual scalar checks.
4.  **Behavioral Analysis:** Generate minimalist visualizations to observe the function's geometry and test how it responds to extreme inputs or outliers.

---

### Technical Constraints

*   **Visualizations:** Use standard Matplotlib. No aesthetic padding or dynamic centering unless it serves the mathematical lesson.
*   **Code Style:** Prioritize mathematical exactness using NumPy and SymPy. Fix specific logic errors without adding software-style "clean code" bloat or unnecessary error handling.
*   **Tone:** Blunt, concise, and focused on the math. No mentorship advice, hollow praise, or conversational filler.
