# 1.Simple Plotter

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 2*np.pi, 500)

x = 3 * np.cos(t)
y = 5 * np.sin(t)

plt.plot(x, y)
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.show()
```

# 2. Sympy reconciliation

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

t = sp.symbols('t')

# symbolic parameterization
x_sym = 3 * sp.cos(t)
y_sym = 5 * sp.sin(t)

# convert symbolic expressions into numpy functions
x_func = sp.lambdify(t, x_sym, 'numpy')
y_func = sp.lambdify(t, y_sym, 'numpy')

# numerical values
t_vals = np.linspace(0, 2*np.pi, 500)

x_vals = x_func(t_vals)
y_vals = y_func(t_vals)

plt.plot(x_vals, y_vals)
plt.grid()
plt.axhline(0)
plt.axvline(0)
plt.show()
```

# 3. by the formal equation

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 500)
y = np.linspace(-7, 7, 500)

X, Y = np.meshgrid(x, y)
# ellipse equation
Z = (X**2 / 9) + (Y**2 / 25)

plt.contour(X, Y, Z, levels=[1])
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.show()
```
