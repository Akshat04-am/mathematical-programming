```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(dpi=300, figsize=(4,2))

doc1 = np.array([[40],
                [15]])
doc2 = np.array([[50],
                [25]])

plt.quiver(0,0, 
           doc1[0], doc1[1],
            angles='xy',
            scale_units='xy',
            scale=1,
            color="yellow",
            label="doctor 1")

plt.quiver(0,0, 
           doc2[0], doc1[1],
            angles='xy',
            scale_units='xy',
            scale=1,
            color="red",
            label="doctor 2")

plt.xlim(0,60)
plt.ylim(0,30)
plt.axvline(0, color="black")
plt.axhline(0, color="black")
plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.figure(dpi=300, figsize=(4,2))

exe1 = np.array([[15],
                [-40]])
exe2 = np.array([[25],
                [-50]])

doc1 = np.array([[40],
                [15]])
doc2 = np.array([[50],
                [25]])

plt.quiver(0,0, 
           doc1[0], doc1[1],
            angles='xy',
            scale_units='xy',
            scale=1,
            color="yellow",
            label="doctor 1")

plt.quiver(0,0, 
           doc2[0], doc1[1],
            angles='xy',
            scale_units='xy',
            scale=1,
            color="green",
            label="doctor 2")

plt.quiver(0,0, 
           exe1[0], exe1[1],
            angles='xy',
            scale_units='xy',
            scale=1,
            color="red",
            label="executioner 1")

plt.quiver(0,0, 
           exe2[0], exe1[1],
            angles='xy',
            scale_units='xy',
            scale=1,
            color="grey",
            label="executioner 2")

plt.xlim(0,100)
plt.ylim(-100,100)
plt.axvline(0, color="black")
plt.axhline(0, color="black")
plt.legend()
plt.grid()
plt.show()
```
