```python
'''1. **How do you implement a piecewise function on a NumPy array?**
   Example: if (x<0), use (x^2); if (0\le x<2), use (2x+1); otherwise use (\sqrt{x}).
   This forces masking and branching.'''
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-20,20,1000)

mask1 = x < 0
valid1 = x[mask1]
eq1 = np.square(valid1)


mask2 = (x >= 0) & (x < 2)
valid2 = x[mask2]
eq2 = np.sqrt(valid2)


mask3 = x > 10
valid3 = x[mask3]
eq3 = np.log(valid3)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 🔧 Input range
x = np.linspace(-20, 20, 1000)

# 🔧 Output array (same shape as x)
y = np.zeros_like(x)

# 🔧 Piecewise masks
mask1 = x < 0
mask2 = (x >= 0) & (x < 2)
mask3 = x >= 2

# 🔧 Apply piecewise definitions
y[mask1] = x[mask1]**2          # x^2
y[mask2] = 2*x[mask2] + 1      # 2x + 1
y[mask3] = np.sqrt(x[mask3])   # sqrt(x)

# 🔧 Plot
plt.figure()
plt.plot(x, y)
plt.title("Piecewise Function")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-20, 20, 1000)

y = np.where(x < 0, x**2,
    np.where(x < 2, 2*x + 1,
        np.sqrt(x)))

plt.plot(x, y)
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 1000)

y = np.zeros_like(x)
y[x >= 0] = 5

plt.plot(x, y)
plt.title("Discontinuity (Jump)")
plt.grid()
plt.show()
```
