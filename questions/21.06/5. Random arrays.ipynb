## 5. Random arrays

Create a 10×10 array of random numbers.

Find:
1. Maximum value and its position.
1. Minimum value and its position.
1. Mean, median, and standard deviation.


```python
import numpy as np

arr = np.random.randint(0, 100, (10, 10))
print(arr)
```

```python
print(arr.max())
print(arr.argmax())
```

```python
print(np.mean(arr))
print(np.median(arr))
print(np.std(arr))
```

## 6. Boolean masking 
Given:

1. arr = np.random.randint(1, 100, 30)
2. Extract all numbers greater than 50.
2. Replace all numbers less than 25 with 0.
2. Count how many numbers are divisible by 3.

```python
import numpy as np

arr = np.random.randint(1, 100, 30)
print(arr)

# 1. Extract all numbers greater than 50
greater_than_50 = arr[arr > 50]
print(greater_than_50)

# 2. Replace all numbers less than 25 with 0
arr[arr < 25] = 0
print(arr)

# 3. Count numbers divisible by 3
count = np.sum(arr % 3 == 0)
print(count)
```

```python
import numpy as np

arr = np.random.randint(1, 21, size=20)
print("Original:", arr)

# Sort ascending
arr_sorted = np.sort(arr)
print("Ascending:", arr_sorted)

# Sort descending
arr_desc = arr_sorted[::-1]
print("Descending:", arr_desc)

# Top 5 largest elements
print("Top 5:", arr_desc[:5])
```

```python
top5 = arr_sorted[-5:][::-1]
print(top5)
```
