### 1. Find every number between 1 and 10,000 whose digits sum to 15.

```python
lst = []
for i in range(10000):
    total = 0
    n = i
    while n > 0:
        # returns remainder
        digit = n % 10
        total += digit
        # returns quotient.
        n //= 10
        
    if total == 15:
        lst.append(i)
print(lst)
```

```python
# It gives the remainder.

print(10 % 3)   # 1
print(15 % 4)   # 3
print(20 % 5)   # 0

# gives the whole-number quotient (the result left after ignoring the remainder).

print(1000 // 30)   # 3
print(1500 // 4)   # 3
print(12345 // 100)  # 1234
```

```python
# using numpy.
import numpy as np

arr = np.arange(10000)

digit_sum = (
    arr % 10 +
    (arr // 10) % 10 +
    (arr // 100) % 10 +
    (arr // 1000) % 10
)

result = arr[digit_sum == 15]

print(result)
```
