```python
# Find every number between 1 and 10,000 whose digits sum to 15.

lst = []
for i in range(10000):
    total = 0
    n = i
    while n > 0:
        digit = n % 10
        total += digit
        n //= 10
        
    if total == 15:
        lst.append(i)
print(lst)
```
