```python
import matplotlib.pyplot as plt
import numpy as np

c_discrete = np.arange(0, 16)
c_cont = np.linspace(0, 15, 200)

x_discrete = c_discrete
y_discrete = np.zeros_like(c_discrete)

x_cont = c_cont
y_cont = np.zeros_like(c_cont)

plt.scatter(x_discrete, y_discrete, label="whole numbers")
plt.plot(x_cont, y_cont, label="continous")

plt.grid(True)
plt.legend()
plt.axis("equal")

plt.show()
```
