```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10,10,1000)
y = np.full_like(x, np.nan)

mask = x < 0
mask2 = np.max(x >= 0)

y[mask] = x[mask]**2
y[mask2] = x[mask2] + 1

plt.plot(x,y)
```
