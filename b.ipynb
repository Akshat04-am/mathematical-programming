# Question 1 — Two-stage transformation

Compute:

[
y = x^2 - 4
]

Then:

* if (y < 0), take absolute value
* otherwise divide by 2

Plot final result.

# Question 2 — Domain filtering + second-stage logic

Compute:

[
y = \log(x)
]

ONLY where valid.

Then:

* if (y > 1), square it
* otherwise add 3

Plot result while clearly marking invalid regions.

# Question 3 — Piecewise pipeline

Build:

[
y=
\begin{cases}
x^2 & x<0 \
2x+1 & x\ge0
\end{cases}
]

Then create a second output:

* if (y > 10), cap it at 10
* otherwise leave it unchanged

Plot both:

* original (y)
* capped version

This is important because it introduces:

> “post-processing transformations”

Very common in finance/risk systems.

---

# Question 4 — Threshold crossing logic

Compute:

[
y = \sin(x)
]

Then:

* wherever (y > 0.8), replace value with 1
* wherever (y < -0.8), replace value with -1
* otherwise set to 0

This is basically:

> signal generation logic

---

# Question 5 — Payoff style logic

Create:

[
y = x - 3
]

Then:

* if (y < 0), set payoff to 0
* otherwise keep y

Plot:

* payoff
* vertical line at strike (x=3)

This is a basic call option payoff disguised as a coding exercise.
