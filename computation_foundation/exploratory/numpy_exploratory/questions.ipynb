4. **How do you simplify an expression numerically without blindly trusting algebraic cancellation?**
   Example:
   [
   f(x)=\frac{x^2-1}{x-1}
   ]
   But handle (x=1) carefully. Don’t just divide and pretend the issue vanished.

5. **How do you compare two implementations of the same function and detect where they disagree?**
   Example: one version uses NumPy directly, the other uses `lambdify`.
   You need a max error check, but also a way to inspect the bad points.

6. **How do you compute a function only on “good” points while preserving an aligned output array?**
   Example:
   input array has bad values, and your output must stay the same length as input.
   Think masking + placeholder values.

7. **How do you apply transformations in the correct order when the order changes the result?**
   Example: compare
   [
   (x-2)^2 + 3
   ]
   versus
   [
   (x^2 - 2) + 3
   ]
   Same symbols, different logic. That is where weak thinking shows up.

8. **How do you build a function that behaves differently across regions and then plot it cleanly?**
   Example: one formula for negative x, another for positive x.
   The challenge is not the math. It is the branching.

9. **How do you detect and handle discontinuities in a function before plotting or using it downstream?**
   Example:
   [
   f(x)=\frac{1}{x-2}
   ]
   You need to avoid garbage near the singularity and decide whether to mask, clip, or split the domain.

10. **How do you write code that retries with a fallback rule when the first rule fails?**
    Example: try `log(x)`, but if `x <= 0`, replace with a fallback like `nan` or a filtered subset.
    This is basic error-handling logic, which matters more than people admit.

12. **How do you process an array, split it into categories, apply different rules to each category, and then merge the result back?**
    This is the real algorithmic jump: classify → transform → recombine.
