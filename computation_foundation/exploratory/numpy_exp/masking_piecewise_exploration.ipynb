Compute:

[
y = x^2 - 4
]

Then:

* if (y < 0), take absolute value
* otherwise divide by 2

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(-5,5,1000)

eq = x**2 - 4

# if zero then abs, else /2
y = np.where(eq < 0, np.abs(eq), eq/2)

plt.plot(x, eq, label='original')
plt.plot(x, y, label='transformed')
plt.legend()
plt.show()
```

Compute:

[
y = \log(x)
]

ONLY where valid.

Then:

* if (y > 1), square it
* otherwise add 3

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 1000)

# Same shape as x and undefined
y = np.full_like(x, np.nan)

# Valid domain for log
valid_mask = x > 0

# 🔧 First stage
y[valid_mask] = np.log(x[valid_mask])

# 🔧 Second-stage masks (based on y, not x)
square_mask = y > 1
add_mask = y <= 1

# 🔧 Apply second-stage rules
y[square_mask] = y[square_mask]**2

y[add_mask] = y[add_mask] + 3

# 🔧 Plot
plt.plot(x, y)

plt.grid()
plt.show()
```

Build:

[
y=
\begin{cases}
x^2 & x<0 \
2x+1 & x\ge0
\end{cases}
]

Then create a second output:

* if (y > 10), cap it at 10
* otherwise leave it unchanged

Plot both:

* original (y)
* capped version

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10,10,1000)
y1 = np.where(x < 0, x**2, 2*x + 1)

# Create capped copy
y2 = y1.copy()
cap = y2 > 10

y2[cap] = 10

plt.plot(x, y1, label='Original')
plt.plot(x, y2, label='Capped')
plt.legend()
plt.grid()
plt.show()
```

Compute:

[
y = \sin(x)
]

Then:

* wherever (y > 0.8), replace value with 1
* wherever (y < -0.8), replace value with -1
* otherwise set to 0

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2*np.pi, 500)
y = np.zeros_like(x)

y1 = np.sin(x)
y2 = np.where(y1 > 0.8, 1, 
        np.where(y1 < -0.8, -1, 0))

plt.plot(x,y1,label="regular")
plt.plot(x,y2,label="contorted")
plt.legend()
plt.grid()
plt.show()
```
