#### Gayle is thinking of starting her own business. 
Total fixed costs are $19,000 per month and unit variable costs are estimated at $37.50. 

From some preliminary studies that she completed, she forecasts sales of 1,400 units at $50 each, 1,850 units at $48 each, 2,500 units at $46 each, and 2,750 units at $44 each. What price would you recommend Gayle set for her products?

```python
import numpy as np

# vc = variable costs.
fixed_costs = 19000
unit_vc = 37.50

prices = np.array([50, 48, 46, 44])
quantities = np.array([1400, 1850, 2500, 2750])

revenue = prices * quantities
total_vc = unit_vc * quantities
total_costs = fixed_costs + total_vc
profits = revenue - total_costs

for i in range(len(prices)):
    print(f"Price: ${prices[i]} | Profit: ${profits[i]:,.2f}")

# finds the index  of the highest value in an array, rather than the value.
best_idx = np.argmax(profits)
print(f"\nOptimal Price: ${prices[best_idx]}")
print(f"Max Profit: ${profits[best_idx]:,.2f}")
```
